### Question 1 — Line Plot  `Matplotlib`

Calculate the **average marks** of all students in each of the four tests (`Test_1` to `Test_4`). Create a **line plot** to show how the class average changes from test to test. Add a title and axis labels.

```python
# ✏️ Your code here

test_columns = ['Test_1', 'Test_2', 'Test_3', 'Test_4']
test_averages = df[test_columns].mean()


plt.figure(figsize=(8, 5))
plt.plot(test_averages.index, test_averages.values, color='red', marker='o', )


plt.title('Class Average Marks ')
plt.xlabel('Tests')
plt.ylabel('Average Marks')
plt.grid('True')


plt.show()
```

### Question 8 — Heatmap  `Seaborn`

Create a **heatmap** of the correlation between `Study_Hours`, `Attendance`, `Sleep_Hours` and `Marks`. Show the values on the heatmap. Which column has the strongest relationship with `Marks`?

```python
# ✏️ Your code here
numeric_df = df[['Study_Hours', 'Attendance', 'Sleep_Hours','Marks']]
correlation = numeric_df.corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()

print("Study_Hours has the strongest relationship with Marks (correlation = 0.96).")
```
